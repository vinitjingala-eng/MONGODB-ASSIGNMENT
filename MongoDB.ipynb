**6.  What is the role of indexes in MongoDB, and how do they improve performance?**

  -  Indexes in MongoDB are special B-tree or hash data structures that store a small portion of a collection’s data in a sorted format. They allow the database to locate and retrieve specific documents directly without performing a time-consuming full collection scan, drastically accelerating read, filter, sort, and aggregation operations.


     How Indexes Improve Performance :

     • Eliminates Collection Scans: Without an index, MongoDB must read every single document in a collection to check for a match. Indexes provide a direct pointer to the target documents, reducing the number of examined documents to a tiny fraction.

     • Accelerates Sorting: Indexes store field values in a pre-sorted order. When a query requests sorted results, MongoDB can traverse the index sequentially rather than loading and sorting documents in memory.

     • Optimizes Range Queries: For operators like <, >, <=, and >=, an ordered index allows the query planner to jump straight to the start of the range and read outward.

     • Enables Covered Queries: If all fields requested by a query exist inside the index itself, MongoDB can fulfill the query directly from the index structure without ever accessing the underlying source documents (totalDocsExamined is 0).

**7.   Describe the stages of the MongoDB aggregation pipeline?**

  -  The MongoDB aggregation pipeline is a framework used to process and analyze data in a multi-stage sequence. Documents from a collection enter the pipeline, pass through various stages that transform or filter them, and the final output is returned as computed results. The output of one stage serves as the input for the next stage.


  1. Filtering & Ordering Stages :

     • $match: Filters documents to allow only those that meet specific criteria to pass to the next stage. Placing this stage as early as possible optimizes performance by leveraging indexes and reducing data volume early on.

     • $sort: Orders documents based on one or more fields in ascending (1) or descending (-1) order.

     • $limit: Restricts the number of documents passed to the subsequent stage.

     • $skip: Skips a designated number of documents, passing the remaining dataset forward (highly useful for pagination).


  2. Grouping & Reshaping Stages :

     • $group: Groups input documents by a specified identifier expression (_id) and applies accumulator expressions (such as $sum, $avg, $max, $min) to calculate aggregated values for each group.

     • $project: Reshapes documents by explicitly including (1) or excluding (0) fields, injecting newly computed fields, or renaming existing properties.

     • $addFields / $set: Appends new fields or updates existing fields within the document without completely rebuilding its structure like $project.

      • $unset: Removes or excludes specific fields from the documents.

      • $unwind: Deconstructs an array field from the input documents, creating a separate output document for every individual element within that array.


  3. Joining & Writing Stages :

     • $lookup: Performs a left outer join to pull in matching documents from another collection within the same database.

     • $out: Writes the final results of the pipeline directly into a new or existing collection, overwriting its contents.

      • $merge: Merges the final results into a target collection, offering advanced handling for matched and unmatched documents (available in modern MongoDB versions).

**8.   What is sharding in MongoDB? How does it differ from replication?**

  -  In MongoDB, sharding horizontally partitions a single dataset across multiple servers to scale storage and throughput. Replication duplicates the entire dataset across multiple nodes in a replica set to provide high availability, data redundancy, and automatic failover.


     What is Sharding?

     • Purpose: Handles massive datasets and high write/read workloads that exceed the limits of a single server.

     • How it works: Divides collections into smaller chunks using a shard key.

     • Key Components:

	    • Shards: Individual servers or replica sets that store a subset of the data.

	    • Config Servers: Store metadata and the cluster's chunk-to-shard mapping.

	    • Query Routers (mongos): Receive client requests and route them to the correct shard.


     What is Replication?

     • Purpose: Ensures fault tolerance, disaster recovery, and continuous availability if a node crashes.

     • How it works: Maintains an identical copy of the full dataset across a group of nodes called a replica set.

     • Key Components:

	    • Primary Node: Receives all write operations and logs them to an oplog.

	    • Secondary Nodes: Copy the oplog to replicate data and can handle read operations.

	    • Automatic Failover: If the primary goes down, secondaries elect a new primary.

**10.  What are the ACID properties in the context of MongoDB transactions?**

  -  MongoDB is fully ACID-compliant and natively guarantees Atomicity, Consistency, Isolation, and Durability for data operations.

     Historically, MongoDB only offered ACID guarantees at the single-document level. However, since version 4.0, it supports multi-document transactions for replica sets, and version 4.2 extended this to distributed multi-document transactions across sharded clusters.

     How MongoDB implements each of the four ACID properties:

  1. Atomicity ;

     Atomicity guarantees an "all-or-nothing" execution. If any operation inside a transaction fails, the entire transaction is rolled back, ensuring no partial data updates occur.

     • Single-Document: Any write operation to a single document is inherently atomic.

     • Multi-Document: For operations spanning multiple documents, collections, or shards, MongoDB manages state transitions (Active → Committed/Aborted) using sessions, executing rollbacks automatically if an error arises.


  2. Consistency :

     Consistency ensures that a transaction always transitions the database from one valid state to another, strictly adhering to all defined rules, schemas, and constraints.

     • MongoDB enforces consistency using document schema validation rules, unique indexes, and triggers.

     • In a distributed cluster, MongoDB uses read and write concerns (such as linearizable read concern and "majority" write concern) to guarantee that all nodes reflect a consistent state of truth and prevent applications from reading stale data.


  3. Isolation :

     Isolation determines how concurrent transactions interact and ensures they do not interfere with one another.

     • MongoDB uses Snapshot Isolation. When a transaction begins, it operates on a consistent snapshot of the data.

     • Uncommitted writes performed inside a transaction remain completely invisible to external sessions until the transaction is successfully committed. This safely prevents "dirty reads".


  4. Durability :

     Durability ensures that once a transaction commits, its modifications are permanently recorded and will survive system crashes, restarts, or power failures.

     • MongoDB achieves durability through its WiredTiger storage engine journal (Write-Ahead Logging / Oplog).

     • To maximize durability in a distributed environment, transactions should be committed with a w: "majority" write concern, which mandates that data must be safely written to the journals of a majority of replica set nodes before acknowledging success to the client.

**12.   How does MongoDB handle schema validation?**

  -  MongoDB handles schema validation by allowing you to enforce data structure, required fields, and data types at the collection level while still preserving its core flexible schema model.


  1. How Validation Rules Are Defined :

     Validation rules are typically established when creating a new collection using db.createCollection() or added to an existing collection using the collMod command. You can define rules in two ways:

     • JSON Schema (Recommended): MongoDB utilizes the $jsonSchema operator (supporting JSON Schema Draft 4) to specify required fields, allowed BSON data types, and value constraints (e.g., maximum/minimum values, regular expression patterns, or enum lists).

     • Query Operators: For simpler use cases, you can use standard MongoDB query expressions (such as $gt, $in, or $type) to filter out invalid documents.


  2. Validation Enforcement Settings :

     MongoDB provides two primary configuration properties—validationAction and validationLevel—that give you fine-grained control over how strictly rules are applied.

     Validation Action (How violations are handled):

     error (Default) : Hard Validation: MongoDB completely rejects the insert or update operation, throws a validation error, and prevents bad data from entering the collection.

     warn : Soft Validation: MongoDB logs a warning message to the server logs but still allows the invalid document to be saved. Useful for testing or debugging.

     Validation Level (Which documents are checked):

     strict (Default) : MongoDB applies validation rules to all inserts and updates.

     moderate : Validation applies to all new inserts. For updates, it only checks existing documents that already meet the validation rules. Existing non-compliant documents are ignored unless an update violates the schema further.

     off : Schema validation is completely disabled for the collection.


  3. Key Behavioral Traits & Restrictions :

     • No Retroactive Checks: When you apply a new schema validator to an existing collection, MongoDB does not automatically check or fix documents that are already stored.

     • System Exceptions: Schema validation cannot be applied to special internal system collections, such as the admin, local, config, or system databases.

**13.  What is the difference between a primary and a secondary node in a replica set?**

  -  A primary node in a replica set handles all write operations as the single source of truth, while secondary nodes replicate the primary's data asynchronously to provide redundancy and read scaling.

     Key Differences :

     • Write Operations :

	 • Primary Node: The only node that accepts and processes write operations (inserts, updates, and deletes).

	 • Secondary Nodes: Cannot accept write operations directly from client applications.

     Read Operations :

	   • Primary Node: Handles read operations by default.

	  • Secondary Nodes: Can also handle read operations if client applications are configured with specific read preferences.

     Data Synchronization :

	  • Primary Node: Records every data change into its operation log (oplog) and serves as the sync source.

	  • Secondary Nodes: Continuously copy the primary's oplog and apply those changes asynchronously to maintain an identical local copy of the data set.

     Role in Elections and Failover :
     
     • Primary Node: Serves actively until it crashes, becomes disconnected, or steps down.

     • Secondary Nodes: Monitor the primary via heartbeats and automatically hold an election to promote one of themselves to the new primary if the current primary fails.

**15.  Explain the concept of embedded documents and when they should be used?**

  -  An embedded document is a record stored directly inside another parent document rather than in a separate collection or table. This approach creates a hierarchical, nested structure within NoSQL databases like MongoDB.


     What Is an Embedded Document?

     • Nesting: Sub-documents or arrays of data reside inside the main document fields (for example, a user document containing an address sub-document).

     • Denormalization: Related information lives together in a single record, removing the need for multi-table joins or secondary lookup queries.

     • Limits: In MongoDB, documents can nest up to 100 levels deep, provided the total size of the parent document stays under 16 MB.


     When Should Embedded Documents Be Used?

     • One-to-One and One-to-Few Relationships: Use embedding when a child entity belongs strictly to a single parent and has a limited, bounded number of items (e.g., a customer profile with a single shipping address).

     • Read-Heavy Workloads: Use embedding when your application almost always fetches the parent and child data together in a single operation.

     • Atomic Operations: Use embedding when you need to update the parent and its related data safely inside a single atomic write operation.

     • Co-located Lifecycles: Use embedding when the child data changes at the same pace as the parent and does not need to be independently queried or shared across multiple records.

**16.  What is the purpose of MongoDB’s $lookup stage in aggregation?**

  -  The purpose of MongoDB's $lookup stage in an aggregation pipeline is to perform a left outer join from a collection in the same database, allowing you to combine and enrich documents with related data from a second collection.


     Key Functions and Features :

     • Left Outer Join: It reads a foreign collection and embeds matching documents into each input document as a new array field.

     • Preserves Unmatched Documents: If no matching document is found in the foreign collection, it still outputs the original document with an empty array for that field.

     • Simplifies Data Modeling: It helps combine data that has been normalized or split across multiple collections due to security, privacy, or performance considerations.


     Main Parameters :

     The basic syntax requires four primary fields:

     • from: The name of the foreign collection to join with.

     • localField: The field from the documents in the input collection.

     • foreignField: The field from the documents in the from collection that corresponds to localField.

     • as: The name of the new array field added to the resulting documents to store the matches.

**19.  How do MongoDB transactions differ from SQL transactions?**

  -  MongoDB and SQL transactions differ primarily in their structural design, performance trade-offs, and how data relationships dictate their usage. While both support multi-statement ACID transactions to guarantee data integrity, they are engineered for fundamentally different database architectures.

     Core Differences

     SQL Transactions :

     Primary Scope :	Multi-row, multi-table updates linked by foreign keys.


     Necessity :	High , Because data is normalized across tables, even simple actions (like an order with items) require a transaction.


     Architecture :	Built natively for monolithic, centralized systems (though distributed SQL exists).


     Performance Overhead :	Low to Moderate. Highly optimized over decades for rapid lock management across tables.


     Isolation Level :	Configurable (e.g., Read Committed, Repeatable Read, Serializable).

     MongoDB Transactions :

     
     Primary Scope : Multi-document, multi-collection updates (often spanning shards).


     Necessity : Low. Single-document updates are inherently atomic. Most workloads are handled using embedded documents instead of multi-document transactions.


     Architecture : Built for distributed systems (replica sets and sharded clusters) using session-based logic.


     Performance Overhead :	High. Multi-document transactions require greater execution overhead and can impact performance if overused.


     Isolation Level : Strictly uses Snapshot Isolation for multi-document operations.

**20.  What are the main differences between capped collections and regular collections?**

 -  The main difference between a capped collection and a regular collection in MongoDB is that a capped collection has a fixed size and acts like a circular buffer, whereas a regular collection has an unlimited size and grows dynamically.


     Key Differences :

     Size and Capacity:

	  • Capped Collections: Have a fixed maximum size in bytes (and optionally a maximum document count) set at creation. When full, they automatically overwrite the oldest documents to make room for new ones.

	  • Regular Collections: Grow dynamically with no preset storage cap. They do not automatically delete old data when space runs out.

     Creation:

	  • Capped Collections: Must be explicitly created using db.createCollection() with capped: true and a size parameter. You cannot create them implicitly by inserting into a non-existent collection.

	  • Regular Collections: Can be created explicitly or implicitly the first time you insert a document into them.

     Deletion and Updates:

	  • Capped Collections: You cannot delete individual documents from them, and updates are only allowed if the new document size does not exceed the old document size (to prevent disk relocation). To clear all data, you must use the emptycapped command.

	  • Regular Collections: Support standard deleteOne(), deleteMany(), and full-size document updates without restriction.

     Insertion and Retrieval Performance:

	  • Capped Collections: Preserve insertion order and store documents in natural disk order. Because they behave like a queue, they do not require extra indexes (even on the _id field) for fast appends, resulting in high insertion and retrieval speed. However, write operations are serialized, reducing concurrent performance.

	  • Regular Collections: Do not guarantee natural disk order for insertion sequences and generally rely on indexes for optimal query performance.

     Sharding and Features:

	  • Capped Collections: Cannot be sharded, are not supported in Stable API V1, and do not support writing inside multi-document transactions.

	  • Regular Collections: Support sharding, transactions, and all standard MongoDB features.

     Alternatives:

	  • Capped Collections: Best suited for high-throughput logging, real-time tailing, and circular buffers.

	  • Regular Collections: Handle general-purpose workloads. If you want automatic expiration on regular collections without a hard disk cap, you can use Time To Live (TTL) indexes instead.

**21.   What is the purpose of the $match stage in MongoDB’s aggregation pipeline?**

  -  The main purpose of the $match stage in MongoDB’s aggregation pipeline is to filter the document stream so that only documents matching specified query conditions pass unmodified to the next stage.


     Key Characteristics & Behavior :

     • Filter Function: It works similarly to the db.collection.find() query syntax, outputting either one document (a match) or zero documents (no match) for each input document.

     • Performance & Optimization: Placing $match as early as possible in the pipeline reduces the number of documents that subsequent, more expensive stages (like $lookup, $group, or $unwind) need to process.

     • Index Usage: When placed at the very beginning of a pipeline, $match can take advantage of database indexes just like a standard find() query.

     • SQL Equivalent:

	  • Early in the pipeline, it acts like a traditional WHERE clause.

	  • Used after a $group stage, it acts like a HAVING clause to filter aggregated results.

**23.  What is MongoDB’s WiredTiger storage engine, and why is it important?**

  -  WiredTiger is the default storage engine for MongoDB—the core software component responsible for managing how data is stored, read, and written in memory and on disk.


     Why WiredTiger Is Important :

     • Document-Level Concurrency: Unlike older engines that used collection-level locks, WiredTiger locks at the document level. Multiple clients can read and write to different documents in the same collection at the same time, vastly improving performance on multi-core CPUs.

     • Built-In Compression: It supports transparent data compression out-of-the-box (using Snappy or Zstd by default) for collections and prefix compression for indexes, which significantly reduces storage footprint and disk I/O costs.

     • Checkpoint-Based Durability and Journaling: It writes a consistent snapshot of the data to disk at regular intervals (checkpoints), and uses a write-ahead journal to ensure fast, reliable crash recovery.

     • Multiversion Concurrency Control (MVCC): It uses MVCC to let concurrent threads view a snapshot of the database at a specific point in time, enhancing transaction isolation and read-write concurrency.

     • Data Security & Integrity: It includes end-to-end data checksums to detect disk corruption, and supports Encryption at Rest in MongoDB Enterprise editions

**9.  Use aggregation to group orders by Region and calculate total sales per region?**

```python
from pymongo import MongoClient
from pprint import pprint

# Connect to MongoDB using your actual connection string and bypass validation
connection_string = "MONGODB_CONNECTION_STRING"
client = MongoClient(connection_string, tls=True, tlsAllowInvalidCertificates=True)

# Select database
db = client["SuperstoreDB"]

# Select collection
collection = db["Orders"]

# Aggregation pipeline
pipeline = [
    {
        "$group": {
            "_id": "$Region",
            "Total_Sales": {"$sum": "$Sales"}
        }
    }
]

# Execute aggregation
result = collection.aggregate(pipeline)

# Display results
print("Total Sales by Region:\n")

for document in result:
    pprint(document)

# Close connection
client.close()
```

```python
from pymongo import MongoClient

# Connect to MongoDB using your actual connection string and bypass validation
connection_string = "MONGODB_CONNECTION_STRING"
client = MongoClient(connection_string, tls=True, tlsAllowInvalidCertificates=True)

# Select database
db = client["SuperstoreDB"]

# Select collection
collection = db["Orders"]

# Fetch distinct Ship Mode values
ship_modes = collection.distinct("Ship Mode")

# Display results
print("Distinct Ship Modes:")

for mode in ship_modes:
    print(mode)

# Close connection
client.close()
```

**11.   Count the number of orders for each category?**

```python
from pymongo import MongoClient
from pprint import pprint

# Connect to MongoDB using your actual connection string and bypass validation
connection_string = "MONGODB_CONNECTION_STRING"
client = MongoClient(connection_string, tls=True, tlsAllowInvalidCertificates=True)

# Select database
db = client["SuperstoreDB"]

# Select collection
collection = db["Orders"]

# Aggregation pipeline
pipeline = [
    {
        "$group": {
            "_id": "$Category",
            "Total_Orders": {"$sum": 1}
        }
    }
]

# Execute aggregation
result = collection.aggregate(pipeline)

# Display results
print("Number of Orders for Each Category:\n")

for document in result:
    pprint(document)

# Close connection
client.close()
```
